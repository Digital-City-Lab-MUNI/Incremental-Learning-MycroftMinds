# Imports

```python
import pandas as pd
import numpy as np
import seaborn as sns
import glob
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

plt.rcParams['figure.figsize'] = [15, 8]
```

# Load data

```python
DATA_DIR = '../../data/second_batch/'
CATEGORY = "meter"
```

```python
all_files_meter1 = glob.glob(DATA_DIR + CATEGORY + "1/*.csv")
all_files_meter2 = glob.glob(DATA_DIR + CATEGORY + "2/*.csv")


all_files = all_files_meter1 + all_files_meter2
df_list = []

for filename in all_files:
    df = pd.read_csv(filename, index_col=0, header=0)
    # data format is 5/6/24 22:03:04
    df['timestamp'] = pd.to_datetime(df['ts'], unit='s')
    df = df.drop(columns=['datum', 'ts', 'granularity', 'pin']) 
    df = df.rename(columns={'spotreba [kWh]': 'consumption'})
    df_list.append(df)


df = pd.concat(df_list, axis=0, ignore_index=False)
df = df.reset_index(names='site_id')
del df_list
df
```
